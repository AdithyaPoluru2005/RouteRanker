```python

import pandas as pd
import os

base_path = r"D:\RouteRanker\data\processed"

cities = [
    "delhi",
    "hyderabad",
    "bengaluru",
    "chennai",
    "mumbai",
    "pune"
]

city_data_m4 = {}

for city in cities:
    city_path = os.path.join(base_path, city)

    city_data_m4[city] = {
        "route_profile": pd.read_csv(
            os.path.join(city_path, "route_profile.csv")
        ),
        "hourly_trips": pd.read_csv(
            os.path.join(city_path, "hourly_trip_counts.csv")
        ),
        "time_features": pd.read_csv(
            os.path.join(city_path, "time_features.csv")
        )
    }

    print(f"\n{city.upper()}")
    for name, df in city_data_m4[city].items():
        print(name, ":", df.shape)

print("\nAll six cities loaded successfully!")
```

```python

for city in cities:
    print("\n" + "=" * 50)
    print(city.upper())
    print("=" * 50)

    for name, df in city_data_m4[city].items():
        print(f"\n--- {name.upper()} ---")

        print("Columns:", df.columns.tolist())

        print("\nData types:")
        print(df.dtypes.to_dict())

        print("\nMissing values:")
        missing = df.isnull().sum()
        print(missing[missing > 0])

        print("\nTotal missing values:", df.isnull().sum().sum())
        print("Duplicate rows:", df.duplicated().sum())
```

```python

quality_summary = []

for city in cities:
    for name, df in city_data_m4[city].items():

        missing_cols = df.isnull().sum()
        missing_cols = missing_cols[missing_cols > 0]

        quality_summary.append({
            "City": city.title(),
            "Dataset": name,
            "Rows": len(df),
            "Columns": len(df.columns),
            "Missing Values": df.isnull().sum().sum(),
            "Duplicate Rows": df.duplicated().sum(),
            "Missing Columns": ", ".join(missing_cols.index)
        })

quality_df = pd.DataFrame(quality_summary)

display(quality_df)
```

```python
for city in cities:
    df = city_data_m4[city]["route_profile"]

    missing_both = df[
        df["total_trips"].isnull() & df["total_stops"].isnull()
    ]

    missing_one = df[
        df["total_trips"].isnull() != df["total_stops"].isnull()
    ]

    print(
        city.upper(),
        "| Both missing:", len(missing_both),
        "| Only one missing:", len(missing_one)
    )

    if len(missing_both) > 0:
        display(missing_both.head(3))
```

```python
for city in cities:
    profile = city_data_m4[city]["route_profile"]
    hourly = city_data_m4[city]["hourly_trips"]

    missing = profile[profile["total_trips"].isnull()]

    if len(missing) > 0:
        profile_ids = set(missing["route_id"].astype(str))
        hourly_ids = set(hourly["route_id"].astype(str))

        matched = profile_ids.intersection(hourly_ids)

        print(
            city.upper(),
            "| Missing routes:", len(missing),
            "| Found in hourly data:", len(matched),
            "| Not found:", len(profile_ids - hourly_ids)
        )
```

```python
import copy

cleaned_data = copy.deepcopy(city_data_m4)

for city in cities:
    df = cleaned_data[city]["route_profile"]

    df["total_trips"] = df["total_trips"].fillna(0)
    df["total_stops"] = df["total_stops"].fillna(0)

    print(
        city.upper(),
        "| Remaining missing values:",
        df[["total_trips", "total_stops"]].isnull().sum().sum()
    )
```

```python
model_data = {}

for city in cities:
    df = cleaned_data[city]["route_profile"]

    model_data[city] = df[
        ["total_trips", "total_stops"]
    ].copy()

    print(city.upper(), model_data[city].shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaled_data = {}

for city in cities:
    scaler = StandardScaler()

    scaled_array = scaler.fit_transform(model_data[city])

    scaled_data[city] = pd.DataFrame(
        scaled_array,
        columns=["total_trips_scaled", "total_stops_scaled"]
    )

    print(city.upper(), scaled_data[city].shape)
```

```python
import os

output_path = r"D:\RouteRanker\data\model_ready"

os.makedirs(output_path, exist_ok=True)

for city in cities:

    final_df = pd.DataFrame({
        "route_id": cleaned_data[city]["route_profile"]["route_id"],
        "total_trips": model_data[city]["total_trips"],
        "total_stops": model_data[city]["total_stops"],
        "total_trips_scaled": scaled_data[city]["total_trips_scaled"],
        "total_stops_scaled": scaled_data[city]["total_stops_scaled"]
    })

    file_path = os.path.join(output_path, f"{city}_processed.csv")

    final_df.to_csv(file_path, index=False)

    print(city.upper(), "saved successfully")

print("\nAll processed datasets saved!")
```

# Module 4: Data Preprocessing

## 1. Objective
The objective of this module was to prepare the feature-engineered public transportation data for machine learning.

## 2. Data Inspection
- Loaded datasets for Delhi, Hyderabad, Bengaluru, Chennai, Mumbai and Pune.
- Checked data types, missing values and duplicate records.
- No duplicate rows were found in the inspected datasets.

## 3. Missing Value Handling
- Delhi, Hyderabad and Pune had no missing values in the selected numerical features.
- Bengaluru, Chennai and Mumbai had missing values in `total_trips` and `total_stops`.
- The affected routes were not found in the corresponding hourly trip datasets.
- Their missing values were replaced with zero to represent zero observed trips and stops in the collected data.
- Delhi's `route_short_name` was excluded because it was completely missing.

## 4. Feature Selection
Selected features:
- `total_trips`
- `total_stops`

Route IDs and route names were excluded from model features because they are identifiers, not numerical measurements.

## 5. Categorical Encoding
Categorical encoding was not required because the selected model features were numerical. Categorical identifiers and route names were excluded.

## 6. Feature Scaling
StandardScaler was applied to the selected numerical features separately for each city to bring them to a comparable scale.

## 7. Training and Validation
A conventional training and validation split was not performed because the prepared dataset is intended for unsupervised clustering, which does not require target labels.

## 8. Output
Six processed CSV files were saved in the `data/model_ready` folder. The original datasets were preserved.

## 9. Limitations
- The selected features represent scheduled trip and stop information, not actual passenger demand.
- Zero values represent no observations in the collected data and do not prove that a route has no real-world service.
- Only route-level numerical features were selected for the initial clustering preparation.
